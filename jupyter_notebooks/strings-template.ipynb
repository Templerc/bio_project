### Store and print a variable

```python
# define some strings
first = 'bob'
last = 'dylan'

# concatenate those strings
full = first + last

# print the strings
print(first)
print(last)
print(full)
print(first + last + "first")
print(first + ' ' + last + ' is the queen')
```

### Get the length of a string

```python
# the long (but readable) way
name_length = len(full) # should be 8
print(name_length)


# a more concise way to do it, functions within functions
print(len(full))
```

### Print strings and numbers

```python
# print strings and integers
print("There are " + str(name_length) + " characters in Bob Dylan's name.")
print("There are", str(name_length), "characters in Bob Dylan's name.")

# formatted print
print(f"There are {name_length} characters in Bob Dylan's name.")
```

### If x = 7 and y = "7", write two print statements that use `x` and `y` to print "77" and "14" as the output

```python
x = 7
y = "7"

# print 77 
print(str(x) + y)
# print 14
print(x + int(y))
```

## Some operations on strings

```python
# methods to change the case of a string
print(type(first), type(last))
print(first, last)
print(first.upper(), last.upper())
print(first.lower(), last.lower())

# make the name properly cased
db_name = "CALeB"
print(db_name.lower().capitalize())
```

### Problem #1: Write a program that asks for the user's name and age, then prints a personalized greeting using string formatting (`.format`).

For *Andy* and *32*, the output would be:

Hello, Andy! You are 32 years old.


```python
# code here
name = input('Enter your name: ')
age = input("Enter your age: ")
print('Hello, {}! You are {} years old.'.format(name.lower().capitalize(), age))
```

### Problem #2: Ask the user for a big number. If the user enters 1000000 (one million), the output will be:

This sure is a big number: 1,000,000

```python
big_num = input("Enter your big number here: ")
print(type(big_num))

print(f'This sure is a big number: {int(big_num):,d}')
print(f'This sure is a big number: {int(big_num):e}')
```
