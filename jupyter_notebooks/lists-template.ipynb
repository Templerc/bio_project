## Working with lists

```python
# declare an empty string
bob = ''

# declare a list
beatles = []
```

### What kind of variable is this?

```python
print(type(bob))
print(type(beatles))
```

### Populate and print a list

```python
beatles = ["John", "Paul", "George", "Ringo", "Stuart"]

# print the whole list
print(beatles)
```

### Calculate and print the length of a list

```python
# calculate number of beatles
number_of_beatles = len(beatles)
# print number of beatles
print(f"There were {number_of_beatles} members of the Beatles.")
```

### Print an item in a list 
   
   _see PowerPoint slides_

```python
# remind us of the list
print(beatles)

# print the first beatle
print(beatles[0])

# many ways to print the last Beatle
print(beatles[4])
print(beatles[-1])
print(beatles[len(beatles) - 1])

# many ways to print George in the Beatle
print(beatles[2])
print(beatles[-3])
```

### Remove an item from a list

```python
# Remove the last item in the list (bye, Stuart)
stu = beatles.pop()

# What happened to the Beatles?
print(stu)
print(beatles)
```

### Make a new list

```python
# make a list called 'stones' with the Rolling Stones (Mick, Ron, and Keith)
stones = ["Mick", "Keith", "Ron"]
```

### Add two lists

```python
# add two lists together
print(beatles)
print(stones)
print("these two groups separately: {}, {}".format(beatles, stones))

# print our new list
supergroup = beatles + stones

# print new list
print(f"{len(supergroup)} members in our new band: {supergroup}")
```

### Make some new lists

```python
# start fresh with new groups
jackson5 = ['jackie', 'tito', 'jermaine', 'marlon', 'michael']
destiny  = ['beyonce', 'kelly', 'michelle']
top40    = ['taylorswift', 'postmalone']

# remove Michael
jackson5.pop()

# make a supergroup
supersupergroup = jackson5 + destiny + top40
```

### Growing a list with `extend()`

```python
# print our lists
print(f"jackson5: {jackson5}")
print(f"destiny: {destiny}")
print(f"top40: {top40}")
print(supersupergroup)

# use extend() to add Destiny's Child to the Jackson 5
```

```python

# extend() modifies a list in place, it doesn't return a value (i.e., a new list)
jackson5.extend(destiny) # extend: jackson5 = jackson5 + destiny

print(f"Here is the extended jackson5: {jackson5}")
```

```python
# use "+" if you want to create a new list
# jackson5 = jackson5 + destiny
```

### Growing a list with `append()`

```python
# use `append()` to add lists in a different way
num = [1, 2, 3]
let = ['a', 'b', 'c']
print(f"our lists: {num, let}")

# What did append do?
num.append(let)
print(f"after appending let to num: {num}")

more_num = [4, 5, 6]
more_num.append(7)
print(more_num)
i = 8 
more_num.append(i)
print(more_num)
```

```python
# How long is this list?
print(len(num))

# print the letters
print(num[-1])
print(num[3])
```

```python
# find and print b
print(type(num[3]))

# print an item in a list 
print(num[3][1])
```
